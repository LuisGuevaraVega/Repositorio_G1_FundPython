```python
import numpy as np
```

Importo numpy con el alias np, asi lo usa todo el mundo y es mas corto de escribir.

```python
array = np.arange(1, 11)
print(array)
```

arange me crea el rango de numeros. Puse 1 y 11 porque el segundo numero no se incluye, entonces asi me sale del 1 al 10.

```python
print(array.max())
print(array.min())
```

max y min son metodos del array, van pegados con punto y se llaman con parentesis vacios.

```python
ceros = np.zeros(5)
print(ceros)
```

zeros(5) me da un array de 5 ceros. Salen como 0. con punto porque numpy los guarda como decimales por defecto, no es error.

```python
unos = np.ones(5)
print(unos)
```

Lo mismo pero con unos.

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(matrix)
```

Aca ya es una matriz de 2 dimensiones. Cada lista de adentro es una fila: la primera fila es 1,2,3 y la segunda 4,5,6.

```python
print(matrix.shape)
```

shape me dice las dimensiones, primero filas y despues columnas, en este caso 2 filas y 3 columnas. El resultado sale como tupla y no como lista, por lo mismo que se ve en el notebook de tuples: no deberia poder cambiarse por accidente.
