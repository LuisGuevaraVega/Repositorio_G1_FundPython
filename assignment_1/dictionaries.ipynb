```python
population = {
    "Lima": 10092000,
    "Arequipa": 1008000,
    "Trujillo": 919000
}

print(population)
```

Aca cada ciudad es la llave y la poblacion es el valor. A diferencia de una lista, no busco por posicion sino por el nombre de la ciudad.

```python
print(population.keys())
```

keys() me muestra solo los nombres de las ciudades, sin los numeros.

```python
print(population.get("Lima"))
```

get busca el valor a partir de la llave. Tambien podria hacer population["Lima"] pero con get si la llave no existiera no se cae el programa, devuelve None.

```python
population["Cusco"] = 428000
print(population)
```

Para agregar un dato nuevo solo le pongo la llave y el valor, asi nomas se agrega al diccionario.

```python
population.pop("Trujillo")
print(population)
```

pop elimina el par que le indico, en este caso Trujillo.

```python
print(population)
```

Queda el diccionario final con Lima, Arequipa y Cusco, se agrego uno y se quito otro respecto al original.
