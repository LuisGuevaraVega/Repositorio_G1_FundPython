# Part 2 – Tuples [4 pts]

En este notebook trabajaremos con tuplas en Python, creando una estructura de datos básica, consultando sus elementos, aplicando funciones integradas y analizando su inmutabilidad

```python
# Crear la tupla requerida
ages = (20, 25, 22, 30, 28)

# 1. Imprimir la tupla
print("Tupla original:")
print(ages)
```

```python
# 2. Mostrar el primer y el último elemento de la tupla
primer_elemento = ages[0]
ultimo_elemento = ages[-1]

print("Primer elemento:", primer_elemento)
print("Último elemento:", ultimo_elemento)
```

```python
# 3. Mostrar el valor máximo, el valor mínimo y el número de elementos
print("Valor máximo:", max(ages))
print("Valor mínimo:", min(ages))
print("Número de elementos:", len(ages))
```

```python
# 4. Intentar modificar uno de los elementos de la tupla
# La siguiente línea generará un error (TypeError) a propósito:
#ages[0] = 21
```

**Explicación sobre la modificación de la tupla:**

Python **no permite** modificar los elementos de una tupla porque las tuplas son estructuras de datos **inmutables** (*immutable*). 

Esto significa que una vez que una tupla ha sido creada, su contenido no puede ser alterado, reemplazado, ni se le pueden agregar o quitar elementos de forma directa (*in-place*). Si necesitas una colección de datos cuyos elementos puedan cambiar a lo largo del tiempo, debes utilizar una lista (`list`) en su lugar.
