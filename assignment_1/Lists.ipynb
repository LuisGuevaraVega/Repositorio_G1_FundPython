## Parte 1: Lists

```python
# Lista de grados
grades = [15, 18, 12, 20, 16]
```

```python
# 1. Print the original list.
print(grades)
```

```python
# 2. Add the grade 17 using append().
grades.append(17)
print(grades)
```

```python
# 3. Sort the grades from lowest to highest using sort().
grades.sort()
print(grades)
```

```python
'''
4. Display:
- the highest grade;
- the lowest grade;
- the total number of grades.
'''
print(grades[-1])
print(grades[0])
print(len(grades))
```

```python
# 5. Print the final list.
print(grades)
```
